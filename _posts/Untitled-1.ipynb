```python
import pypdf
```

```python
# Load a PDF file
pdf_path1 = r"C:\Users\antho\OneDrive\Documents\Scan2025-02-21_115404.pdf"
pdf1 = pypdf.PdfReader(pdf_path1)

pdf_path2 = r"C:\Users\antho\OneDrive\Documents\Scan2025-02-21_115745.pdf"
pdf2 = pypdf.PdfReader(pdf_path2)
```

```python
from pypdf import PdfWriter

# Create a PdfWriter object
pdf_writer = PdfWriter()

# Add pages from pdf1
for page_num in range(len(pdf1.pages)):
    pdf_writer.add_page(pdf1.pages[page_num])

# Add pages from pdf2
for page_num in range(len(pdf2.pages)):
    pdf_writer.add_page(pdf2.pages[page_num])

# Write out the combined PDF
output_path = r"C:\Users\antho\OneDrive\Documents\Combined.pdf"
with open(output_path, 'wb') as output_pdf:
    pdf_writer.write(output_pdf)

print(f"Combined PDF saved to {output_path}")
```
